| CORN - SALES, MEASURED IN \$ | `sales_usd` | dollars |

| CORN, SILAGE - PRODUCTION, MEASURED IN TONS | `silage_tons` | tons |

## 1.1 Corn Census data for IOWA

```python
iowa_corn_data = pd.read_csv("Iowa_corn_data.csv")
iowa_corn_data.head(10)
```

```python
iowa_corn_data.info()
```

**What next cell shows.** The file is in *long* format (one row per county × census year × data item). Of the 21 columns, many are constant (`Program`, `State`, `Commodity`, …) or entirely empty (`Week Ending`, `Zip Code`, `Region`, `Watershed`), so I keep only the 7 that matter. There are 99 counties (all of Iowa) and 6 census years (1997, 2002, …, 2022).

```python
profile = pd.DataFrame({
    "distinct_values": iowa_corn_data.nunique(dropna=True),
    "missing":         iowa_corn_data.isna().sum(),
    "example":         iowa_corn_data.iloc[0],
})
profile
```

```python
KEEP = ["Year", "Ag District", "County", "County ANSI", "Data Item", "Value", "CV (%)"]
iowa_corn_filter_data = iowa_corn_data[KEEP].copy()
iowa_corn_filter_data.columns = ["year", "ag_district", "county", "county_ansi", "data_item", "value_raw", "cv_raw"]
iowa_corn_filter_data.head(5)
```

```python
for c in ["ag_district", "county", "data_item", "value_raw", "cv_raw"]:
    iowa_corn_filter_data[c] = iowa_corn_filter_data[c].str.strip()
iowa_corn_filter_data["year"] = iowa_corn_filter_data["year"].astype(int)

# "217,044,000" -> 217044000.0 ; "(D)" -> NaN  (errors="coerce" turns non-numeric codes into NaN)
iowa_corn_filter_data["value"]  = pd.to_numeric(iowa_corn_filter_data["value_raw"].str.replace(",", "", regex=False), errors="coerce")
iowa_corn_filter_data["cv_pct"] = pd.to_numeric(iowa_corn_filter_data["cv_raw"], errors="coerce")     # "(L)", "(D)" -> NaN

# #sanity check: the only reason a Value is missing should be the (D) code
withheld = iowa_corn_filter_data["value_raw"].eq("(D)").sum()
print(f"Values withheld with (D): {withheld}  |  Values that became NaN: {iowa_corn_filter_data['value'].isna().sum()}")
assert withheld == iowa_corn_filter_data["value"].isna().sum()

# 5-digit FIPS code (state 19 + county) is the standard key for joining county data later
# Convert 'county_ansi' to string before concatenation
iowa_corn_filter_data["fips"] = "19" + iowa_corn_filter_data["county_ansi"].astype(str)
iowa_corn_filter_data.head()
```

### Recode the Data Item text into short variable names
The file also contains a 5th item (*corn, traditional or Indian*) that is not one of the four measures in this project. A regex mapping both recodes the four I want and drops the rest.

#### used regex

```python
MEASURE_PATTERNS = {
    r"OPERATIONS WITH SALES":         "operations_with_sales",
    r"SALES, MEASURED IN \$":         "sales_usd",
    r"^CORN, GRAIN - PRODUCTION":     "grain_bu",
    r"^CORN, SILAGE - PRODUCTION":    "silage_tons",
}
def to_measure(item):
    for pattern, name in MEASURE_PATTERNS.items():
        if re.search(pattern, item):
            return name
    return None

iowa_corn_filter_data["measure"] = iowa_corn_filter_data["data_item"].map(to_measure)

dropped = iowa_corn_filter_data[iowa_corn_filter_data["measure"].isna()]
print("Rows dropped (not one of the 4 measures):", len(dropped))
print(dropped["data_item"].unique())

nass = iowa_corn_filter_data.dropna(subset=["measure"]).copy()
nass["measure"].value_counts()
```

### Reshape long → wide
For analysis I want **one row per county-year** with one column per measure. Two derived columns are added:
* `sales_per_op` = sales ÷ operations with sales (average \$ per selling farm)
* `sales_per_bu` = sales ÷ bushels of grain produced — a *rough* revenue-per-bushel proxy

```python
MEAS = ["operations_with_sales", "sales_usd", "grain_bu", "silage_tons"]

wide = (iowa_corn_filter_data.pivot(index=["year", "ag_district", "county", "fips"], columns="measure", values="value")
            .reset_index())
wide.columns.name = None
wide = wide[["year", "ag_district", "county", "fips"] + MEAS]

wide["sales_per_op"] = wide["sales_usd"] / wide["operations_with_sales"]
wide["sales_per_bu"] = wide["sales_usd"] / wide["grain_bu"]

print("Shape:", wide.shape, "→ 99 counties × 6 census years =", 99 * 6)
wide.sample(5, random_state=1)
```

## 1.2 PDSI (Palmer Drought Severity Index) — CSV

*PDSI: about 0 = normal, ≤ −2 = drought, ≥ +2 = unusually moist.*


```python
pdsi = pd.read_csv("PDSI.csv", dtype={"Date": str}).rename(columns={"Date": "date", "Value": "pdsi"})
print(pdsi.shape)
pdsi.head()
```

## 1.3 average temperature — XML

```python
import xml.etree.ElementTree as ET

TEMP_XML = '/content/AvTemp.xml'

root = ET.parse(TEMP_XML).getroot()
print("Title:", root.findtext("description/title"), "|", "Units:", root.findtext("description/units"))

temp = (pd.read_xml(TEMP_XML, xpath=".//data", parser="etree", dtype={"date": str})
          .rename(columns={"value": "avg_temp_f"}))
print(temp.shape)
temp.head()
```

## 1.4 Precipitation — JSON

```python
prcp = pd.read_json("PRECIPITATION.json")
# Extract the value from the 'data' column and convert to numeric, coercing errors
prcp["prcp"] = prcp["data"].apply(lambda x: x.get('value') if isinstance(x, dict) else None)
prcp["prcp"] = pd.to_numeric(prcp["prcp"], errors='coerce')

# Drop the 'description' rows which are not actual data points and the original 'data' column
prcp = prcp.drop(columns=["description", "data"]).dropna(subset=["prcp"])

# Rename the index to 'date' and convert to integer
prcp.index.name = "date"
prcp = prcp.reset_index()
prcp["date"] = prcp["date"].astype(int)

print(prcp.shape)
prcp.head()
```

## 1.5 Integrating the three weather feeds
All three are keyed by `YYYYMM` (always month `08`), so a simple join on `date` works. I check that every join is 1-to-1 and add a `year` column to link to the corn data.

### used merging

```python
prcp["date"] = prcp["date"].astype(str) # Convert prcp date to string to match other dataframes
wx = (temp.merge(prcp, on="date", validate="one_to_one")
          .merge(pdsi,   on="date", validate="one_to_one"))
wx["year"] = wx["date"].str[:4].astype(int)

assert wx["date"].str[4:].eq("08").all(), "unexpected non-August month"
assert len(wx) == 30 and wx.notna().all().all()

# classify each with the usual PDSI cut-offs
wx["pdsi_class"] = pd.cut(wx["pdsi"], bins=[-np.inf, -2, 2, np.inf],
                          labels=["Dry (≤ -2)", "Normal (-2 to 2)", "Wet (≥ 2)"], right=False)
wx.head()
```

## 2.3 Statewide totals by census year
Summing the county values gives a statewide picture. (Silage totals exclude the withheld counties, so they are slight under-counts.)

### used group by

```python
state = wide.groupby("year")[MEAS].sum(min_count=1)      # min_count=1 keeps 1997/2002 sales as NaN instead of 0

summary = pd.DataFrame({
    "grain (bn bu)":      state["grain_bu"] / 1e9,
    "sales ($ bn)":       state["sales_usd"] / 1e9,
    "silage (m tons)":    state["silage_tons"] / 1e6,
    "farms w/ sales":     state["operations_with_sales"],
})
summary["grain chg vs prev census %"] = (summary["grain (bn bu)"] / summary["grain (bn bu)"].shift(1) - 1) * 100
summary["sales chg vs prev census %"] = (summary["sales ($ bn)"]  / summary["sales ($ bn)"].shift(1)  - 1) * 100
summary
```

```python
g = summary["grain (bn bu)"]; s = summary["sales ($ bn)"]; f = summary["farms w/ sales"]
print(f"Grain 1997→2022 : {g[2022]/g[1997]-1:+.1%}")
print(f"Grain 2007→2012 : {g[2012]/g[2007]-1:+.1%}   (drought year)")
print(f"Sales 2007→2022 : {s[2022]/s[2007]-1:+.1%}")
print(f"Grain 2007→2022 : {g[2022]/g[2007]-1:+.1%}")
print(f"Farms 2007→2022 : {f[2022]/f[2007]-1:+.1%}")
```

**Interpretation.**
* Grain production grew strongly over the long run (**1.58 → 2.51 billion bu, +59 %** from 1997 to 2022), but **not in a straight line**: 2012 is a clear dip (**−20 %** vs. 2007) before it recovers to a new high of 2.58 bn bu in 2017.
* **Sales tell a different story from volume.** From 2007 to 2022 grain volume rose only about 10 %, yet sales rose **~132 %** (6.8 → 15.8 \$ bn). Sales in 2012 went *up* 73 % even though production fell 20 % — a sign that price, not volume, drives sales (see 2.5).
* The number of farms with corn sales **fell ~9 %** (49,970 → 45,366) while output grew, which is consistent with **consolidation into fewer, larger operations** (acreage isn't in this data, so this is suggestive, not proven).

## 2.4 Geographic concentration (2022)
Is corn production spread evenly across the 99 counties, or concentrated in a few?

Nine of the top 10 are in the northern or middle bands of the state; the only southern one is Pottawattamie (Southwest).

## 2.5 Sales per farm and implied revenue per bushel

```python
econ = pd.DataFrame({
    "sales per farm ($)":          state["sales_usd"] / state["operations_with_sales"],
    "sales per bushel produced ($)": state["sales_usd"] / state["grain_bu"],
}).dropna()
econ
```

**Interpretation.** *Sales ÷ bushels* jumps from ~\$3 (2007, 2017) to ~\$6.3–6.4 (2012, 2022). That tracks the well-known corn price spikes of the 2012 drought and 2022, which is a useful sanity check that the data hang together. **Caveats:** it is a rough proxy — sales may cover corn not counted in grain production, sales in a census year don't line up exactly with that year's harvest, and values are **nominal dollars** (not inflation-adjusted). Sales per farm rose from ~\$136k (2007) to ~\$347k (2022), reinforcing the "fewer, bigger farms" finding.

## 2.9 Linking corn to weather (census years only)
Weather exists every year but corn census data only every 5 years, so the overlap is **n = 6** (n = 4 for sales). This is exploratory: a table is more honest than a fitted line.

```python
link = summary[["grain (bn bu)", "sales ($ bn)"]].join(econ["sales per bushel produced ($)"]).join(
        wx.set_index("year")[["avg_temp_f", "prcp", "pdsi"]])
link
```

```python
link[["grain (bn bu)", "avg_temp_f", "prcp", "pdsi"]].corr().loc["grain (bn bu)"].drop("grain (bn bu)").to_frame("corr with grain (n=6)")
```

**Interpretation.**
* The only census year with a clearly bad August (**2012, PDSI −3.95**) is also the year with the only large fall in production (**−20 %** vs. 2007) *and* the price proxy more than doubled (\$2.97 → \$6.40). That is consistent with drought cutting supply and raising prices.
* Across all six years the correlation of grain with PDSI is only **+0.44**, with rainfall **+0.23**, with temperature **+0.06** — weak, and with n = 6 not statistically meaningful.
* Why not stronger: (1) long-run growth in yields/acres dominates (1997 is the lowest output despite a near-normal August), (2) August is only one slice of the season (pollination is mostly July), and (3) statewide averages hide local conditions. **Takeaway:** weather explains an *outlier year* (2012) better than the general trend.

# Step 3 — Key takeaways, limitations

**Key findings**
1. **Production is up, farms are down.** Grain grew ~59 % (1997–2022) while the number of farms with corn sales dropped ~9 % (2007–2022); sales per farm rose from ≈ \$136k to ≈ \$347k.
2. **Sales follow price more than volume.** 2007→2022 sales +132 % vs. bushels +10 %; implied revenue per bushel flips between ~\$3 and ~\$6.3 in step with known price spikes.
3. **2012 is the outlier year:** driest August in 30 years (PDSI −3.95), grain −20 % statewide, all 9 districts down, then a full rebound by 2017.
4. **Geography:** output is broad-based (top 10 counties ≈ 18 % of grain) but clearly tilted north; silage is a separate, highly skewed livestock-driven pattern.
5. **Weather ↔ corn:** at the census-year level (n = 6) only the 2012 drought stands out; general correlations are weak.

**Limitations**
* Census data come every 5 years → only 6 (or 4) overlap points with weather; no causal claims.
* Weather is a statewide **August-only** average; it ignores the rest of the season and local variation.
* Dollar figures are nominal; sales-per-bushel is a rough proxy.
* `(D)` withheld silage cells (8) are excluded from totals.

## 4.1 2012 drought underperformers, relative to their own district

PS1 found the 2012 drought hit every district, but unevenly. Here we ask a sharper, district-relative question: which counties fell short of **their own district's** typical 2012 output — a fairer bar than one statewide threshold, since districts differ in baseline scale.

### used subset/slice

```python
y12_ps2 = wide[wide["year"] == 2012].copy()
y12_ps2["district_median_grain"] = y12_ps2.groupby("ag_district")["grain_bu"].transform("median")
underperformers = y12_ps2.loc[y12_ps2["grain_bu"] < y12_ps2["district_median_grain"],
                               ["county", "ag_district", "grain_bu", "district_median_grain"]]
print(f"{len(underperformers)} of {len(y12_ps2)} counties fell below their own district's 2012 median")
underperformers.sort_values("ag_district").head(8)
```

**Interpretation.** Close to half (47 of 99) fell below their district's own median — county-level drought impact in 2012 wasn't uniform even *within* a district.

## 4.2 farm_bill_era: census year → federal farm-policy period
Corn economics doesn't move in a vacuum — U.S. farm policy (price supports, crop insurance subsidies, conservation rules) is set by multi-year Farm Bills. Mapping each census year to the Farm Bill in force lets us ask whether growth differed across policy eras — a new, non-geographic lens.

Farm Bills Information:

(https://www.loc.gov/ghe/cascade/index.html?appid=1821e70c01de48ae899a7ff708d6ad8b&bookmark=What%20is%20the%20Farm%20Bil)

### used map/recode

```python
FARM_BILL_ERA = {
    1997: "1996 Farm Bill (1996-2002)",
    2002: "2002 Farm Bill (2002-2008)",
    2007: "2002 Farm Bill (2002-2008)",   # 2008 Farm Bill wasn't enacted until mid-2008
    2012: "2008 Farm Bill (2008-2014)",
    2017: "2014 Farm Bill (2014-2018)",
    2022: "2018 Farm Bill (2018-2023)",
}
wide["farm_bill_era"] = wide["year"].map(FARM_BILL_ERA)
wide.drop_duplicates("year")[["year", "farm_bill_era"]]
```

### used group by

```python
era_order = ["1996 Farm Bill (1996-2002)", "2002 Farm Bill (2002-2008)", "2008 Farm Bill (2008-2014)",
             "2014 Farm Bill (2014-2018)", "2018 Farm Bill (2018-2023)"]
by_era = (wide.groupby("farm_bill_era")
              .agg(census_years=("year", "nunique"),
                   mean_grain_bu=("grain_bu", "mean"),
                   mean_sales_usd=("sales_usd", "mean"),
                   mean_operations=("operations_with_sales", "mean"))
              .reindex(era_order))
by_era
```

**Interpretation.** Average county grain output is lowest under the 1996 Farm Bill (16.0M bu, our earliest year) and highest under the 2014/2018 bills (~26M bu). Average sales per county nearly doubled from the 2002 bill era to the 2018 bill era (\$68.7M → \$159.2M), reflecting price growth more than any single policy change — consistent with PS1's finding that price, not volume, drives sales.

## 4.3 land_value_tier: ag district → land-value

Iowa State's CARD (Center for Agricultural and Rural Development) center publishes an official **\$/acre land value by ag district** every year. Using the published 2022 district averages, we bucket the nine districts into three land-value tiers with an explicit dictionary — a *different* partition from any grouping used in PS1, since it's grouped by actual dollar value rather than compass direction.

https://www.card.iastate.edu/


### used map/recode

```python
# Source: Iowa State University CARD, 2022 ISU Land Value Survey, district averages ($/acre)
DISTRICT_LAND_VALUE_2022 = {
    "NORTHWEST": 14_878, "NORTH CENTRAL": 12_449, "NORTHEAST": 11_627,
    "WEST CENTRAL": 12_411, "CENTRAL": 12_582, "EAST CENTRAL": 12_595,
    "SOUTHWEST": 9_264, "SOUTH CENTRAL": 6_824, "SOUTHEAST": 9_276,
}
def land_value_tier(district):
    v = DISTRICT_LAND_VALUE_2022[district]
    if v >= 12_000:
        return "High land value (>=$12k/acre)"
    elif v >= 9_000:
        return "Mid land value ($9-12k/acre)"
    return "Low land value (<$9k/acre)"

wide["land_value_tier"] = wide["ag_district"].map(land_value_tier)
tier_check = wide.drop_duplicates("ag_district")[["ag_district", "land_value_tier"]]
tier_check["district_$_per_acre"] = tier_check["ag_district"].map(DISTRICT_LAND_VALUE_2022)
tier_check.sort_values("district_$_per_acre", ascending=False)
```

### used groupby

```python
by_tier_year = (wide.groupby(["land_value_tier", "year"])
                     .agg(n_counties=("county", "count"),
                          mean_grain_bu=("grain_bu", "mean"),
                          cv_grain=("grain_bu", lambda s: s.std() / s.mean()))
                     .round(3))
by_tier_year.unstack("land_value_tier")["cv_grain"]
```

**Interpretation.** In **every single census year**, the High land-value tier has the lowest coefficient of variation (~0.25–0.29) — its counties produce fairly similar amounts of corn. The Low tier is far more erratic (0.48–0.70). This makes agronomic sense: in the core corn belt, nearly every acre is good corn ground, so counties look alike; in the low-value tier, some counties have much more corn-suitable land than their neighbors even within the same district. Not just "richer land produces more" — **richer land produces more *predictably*.**

# Step 5: farm operating expenses

PS1 showed that Iowa corn **production grew while the number of corn-selling farms fell** — grain output rose ~59% from 1997 to 2022 even as farms with corn sales dropped ~9% from 2007 to 2022 — consistent with consolidation into fewer, larger operations. This phase extends that story from *how much corn is produced* to **what it costs to produce it, and whether that cost structure tracks land quality**.


Using Expenses data from USDA (https://quickstats.nass.usda.gov/)

Expenses : (fertilizer, chemicals, seed, fuel, cash rent, hired labor, interest, property taxes, and total operating expense)

## 5.1 Read and reshape the expense file


```python
EXPENSES_CSV = "expenses.csv"
exp_raw = pd.read_csv(EXPENSES_CSV)
exp_raw.head(5)
```

```python
EXPENSE_ITEMS = {
    "EXPENSE TOTALS, OPERATING - EXPENSE, MEASURED IN $":                                          "total_opex_usd",
    "EXPENSE TOTALS, OPERATING - OPERATIONS WITH EXPENSE":                                          "total_opex_farms",
    "FERTILIZER TOTALS, INCL LIME & SOIL CONDITIONERS - EXPENSE, MEASURED IN $":                     "fertilizer_usd",
    "CHEMICAL TOTALS - EXPENSE, MEASURED IN $":                                                      "chemical_usd",
    "SEEDS & PLANTS TOTALS - EXPENSE, MEASURED IN $":                                                 "seed_usd",
    "FUELS, INCL LUBRICANTS - EXPENSE, MEASURED IN $":                                                "fuel_usd",
    "RENT, CASH, LAND & BUILDINGS - EXPENSE, MEASURED IN $":                                          "rent_cash_usd",
    "LABOR, HIRED - EXPENSE, MEASURED IN $":                                                          "labor_hired_usd",
    "INTEREST - EXPENSE, MEASURED IN $":                                                              "interest_usd",
    "TAXES, PROPERTY, REAL ESTATE & NON-REAL ESTATE, (EXCL PAID BY LANDLORD) - EXPENSE, MEASURED IN $": "property_tax_usd",
}
e = exp_raw[exp_raw["Data Item"].isin(EXPENSE_ITEMS) & (exp_raw["Domain"] == "TOTAL")].copy()
e = e[["Year", "County", "Data Item", "Value"]]
e.columns = ["year", "county", "data_item", "value_raw"]
e["year"] = e["year"].astype(int)
e["county"] = e["county"].astype(str).str.strip()
e["value"] = pd.to_numeric(e["value_raw"].astype(str).str.replace(",", "", regex=False), errors="coerce")
e["item"] = e["data_item"].map(EXPENSE_ITEMS)

print("rows kept:", len(e), "| non-numeric values after conversion:", e["value"].isna().sum())
```

```python
expense = e.pivot(index=["year", "county"], columns="item", values="value").reset_index()
expense.columns.name = None
print("expense panel:", expense.shape, "(expect 99 x 6 = 594)")
expense.head(5)
```

## 5.2 Check before merging

```python
# (a) Do the two files describe the exact same set of counties?
same_counties = set(wide["county"]) == set(expense["county"])
print("Same 99 county names in both files:", same_counties)

# (b) Is (year, county) a unique key in EACH file separately?
print("wide key unique:",    ~wide.duplicated(["year", "county"]).any())
print("expense key unique:", ~expense.duplicated(["year", "county"]).any())

# (c) Same year coverage?
print("wide years:", sorted(wide["year"].unique()))
print("expense years:", sorted(expense["year"].unique()))
```

## 5.3 Perform the merge
validate="one_to_one" makes pandas **raise an error** if either side turns out to have duplicate keys. indicator=True reports which rows matched.

```python
panel = wide.merge(
    expense,
    on=["year", "county"],
    how="outer",
    validate="one_to_one",
    indicator=True,
)
print("rows before merge: wide =", len(wide), " expense =", len(expense))
print("rows after merge:", len(panel))
panel["_merge"].value_counts()
```

```python
assert len(panel) == len(wide) == len(expense), "row count changed — merge fanned out!"
assert (panel["_merge"] == "both").all(), "some rows failed to match!"
print("Row count unchanged and every row matched. Merge is 1-to-1 and complete, as expected.")

check = panel[(panel["year"] == 2022) & (panel["county"] == "BOONE")]
print(check[["year", "county", "grain_bu", "sales_usd", "total_opex_usd", "fertilizer_usd", "rent_cash_usd"]])
```

## 5.5 flag an unreliable ratio created by the merge

A natural post-merge metric is cost per bushel (`total_opex_usd / grain_bu`). But `total_opex_usd` covers all farm operations reporting any expense, while `grain_bu` covers only corn. If a county's expense-reporting population is much larger than its corn-selling population, the ratio compares apples to a much bigger fruit basket. We compute `opfarm_ratio = total_opex_farms / operations_with_sales` to measure that mismatch, then use `.mask()` to null out `cost_per_bu` wherever the mismatch is large (> 2.5×)

### used replace on condition(.mask)

```python
panel["cost_per_bu"]  = panel["total_opex_usd"] / panel["grain_bu"]
panel["opfarm_ratio"] = panel["total_opex_farms"] / panel["operations_with_sales"]

flagged = panel["opfarm_ratio"] > 2.5
panel["cost_per_bu_reliable"] = panel["cost_per_bu"].mask(flagged)

print(f"{flagged.sum()} of {panel['opfarm_ratio'].notna().sum()} county-years flagged as unreliable")
panel.loc[flagged, "ag_district"].value_counts()
```

**Interpretation.** The 65 flagged county-years are overwhelmingly **South Central (43) and Southeast (17)** — exactly our `land_value_tier` "Low" and "Mid" districts. Those districts have relatively little corn acreage but plenty of pasture/livestock operations that report general farm expenses without selling corn, so `total_opex_farms` there runs 3–5× `operations_with_sales` (vs. ~1.3–2× in the corn belt). Nulling `cost_per_bu` there prevents a misleading "cost per bushel" for places where corn isn't really what most reporting farms do.

## 5.6 High Scale and low cost counties 2022

High-scale (grain_bu at/above the statewide median) AND low-cost(reliable cost-per-bushel at/below the statewide median) -- i.e. counties that are both big producers and cheap-to-run, using the merged panel and the cost_per_bu_reliable column from 5.5 (so the low-cost side excludes counties whose ratio was already flagged as unreliable).

### used subset/slice

```python
y22 = panel[panel["year"] == 2022].dropna(subset=["cost_per_bu_reliable"]).copy()

grain_median = y22["grain_bu"].median()
cost_median  = y22["cost_per_bu_reliable"].median()

y22["efficient"] = (y22["grain_bu"] >= grain_median) & (y22["cost_per_bu_reliable"] <= cost_median)

efficient_counties = y22.loc[y22["efficient"],
                              ["county", "ag_district", "grain_bu", "cost_per_bu_reliable"]
                             ].sort_values("grain_bu", ascending=False)

print(f"{efficient_counties.shape[0]} of {y22.shape[0]} counties (with a reliable cost ratio) "
      f"are both high-scale and low-cost in 2022")
efficient_counties.head(10)
```

24 of 83 counties (those with a reliable cost ratio) clear both bars — output at/above the statewide median (29.8M bu) and cost per bushel at/below the median ($11.37) — led by Kossuth, Pottawattamie, and Buchanan. They're concentrated in Central, West Central, and East Central (7, 5, 5 counties), with Southwest represented by just Pottawattamie alone

## 5.7 Does rent track land value?
This is the payoff the merge was for: `land_value_tier` (Step 4, built from a *published* source) predicts a hypothesis about `rent_cash_usd` (Step 5, from the *merged* row-level data). If our tiering does real work, cash rent per farm operation should rise from Low → Mid → High tier.

```python
rent_by_tier = (panel.groupby("land_value_tier")
                      .apply(lambda d: d["rent_cash_usd"].sum() / d["total_opex_farms"].sum(),
                             include_groups=False)
                      .reindex(["Low land value (<$9k/acre)", "Mid land value ($9-12k/acre)", "High land value (>=$12k/acre)"])
                      .rename("avg_cash_rent_per_operation_$"))
rent_by_tier
```

```python
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.bar(rent_by_tier.index, rent_by_tier.values, color=["#C8C8C8", CORN, "#B23A48"])
ax.set_ylabel("$ cash rent expense / operation")
ax.set_title("Cash rent tracks published land value")
plt.xticks(rotation=12, ha="right")
plt.tight_layout();
plt.show();
```

**Interpretation.** Average cash rent expense per farm operation rises cleanly across tiers: **\$8,338 (Low) → \$18,183 (Mid) → \$30,702 (High)** — roughly a **3.7×** gap between the lowest- and highest-value tiers, tracking the underlying \$6,824–\$14,878/acre spread in the published district land values almost exactly in relative terms. A variable built purely from *district-level published averages* (Step 4) lines up with an independent, *farm-level reported expense* variable brought in by the Step 5 merge. It also strengthens the 4.7 finding — counties paying more for land also produce more consistently, which is exactly what we'd expect if they're farming more uniformly good corn ground.

## 5.8 rent-per-operation outliers

rent_cash_usd / total_opex_farms gives average cash rent paid per farm operation. A few county-years are extreme statistical outliers (|z| > 3) -- not misleading like 5.5's ratio mismatch, just unusually high rent relative to the rest of the panel. np.where() replaces those specific values with NaN in a new column, while keeping the original intact.

```python
panel["rent_per_farm"] = panel["rent_cash_usd"] / panel["total_opex_farms"]

z = (panel["rent_per_farm"] - panel["rent_per_farm"].mean()) / panel["rent_per_farm"].std()
panel["rent_per_farm_clean"] = np.where(z.abs() > 3, np.nan, panel["rent_per_farm"])

outliers = panel.loc[z.abs() > 3, ["year", "county", "ag_district", "rent_per_farm"]]
outliers.sort_values("rent_per_farm", ascending=False)
```

## 5.9 flag abnormal county-over-time fertilizer jumps

Fertilizer spend for the SAME county can look like a data error if it swings wildly between census years -- but it can also be a real price shock. We compare each county-year to that same county's prior census
(via groupby().shift()) and flag ratios above 3x as needing a closer look before treating the raw dollar change as "more fertilizer use."

### used replace on (np.where)

```python
# ---- Replace on condition: flag abnormal county-over-time fertilizer jumps
panel = panel.sort_values(["county", "year"])
panel["fertilizer_prev_census"] = panel.groupby("county")["fertilizer_usd"].shift(1)
panel["fertilizer_yoy_ratio"] = panel["fertilizer_usd"] / panel["fertilizer_prev_census"]

spike = panel["fertilizer_yoy_ratio"] > 3
panel["fertilizer_usd_flagged"] = np.where(spike, np.nan, panel["fertilizer_usd"])

print(f"{spike.sum()} county-years flagged as an abnormal fertilizer jump vs. their prior census")
panel.loc[spike, ["year", "county", "fertilizer_prev_census", "fertilizer_usd", "fertilizer_yoy_ratio"]]
```

Interpretation. Only 2 of 495 comparable county-years trigger: Emmet and Palo Alto, 2002 → 2007, where fertilizer spend roughly tripled (ratio ≈ 3.09x) while every other county's ratio sits under 3x. That's not a data error — it lines up with the well-known mid-2000s run-up in fertilizer/fuel input prices heading into the 2007–2008 commodity price boom, so the honest move here is to note it, not silently null it out and lose a real economic signal.
